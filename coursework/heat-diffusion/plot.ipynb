```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
""" 
Code to make the log-log plot
"""
# read the data from the text file
with open("error.txt") as f:
    data = []
    for line in f.readlines()[1:]:
        line = line.strip().split()
        line = [float(x.strip(',')) for x in line]
        data.append(line)
data = np.array(data)

# extract the errors from the data
dt = data[:,0]
error1 = data[:,1]
error2 = data[:,2]

# create a log-log plot of the two errors
plt.loglog(dt[2:], error1[2:], 'ro', label='Forward Euler Scheme')
plt.loglog(dt, error2, 'g^', label='Backward Euler Scheme')
plt.xlabel('$\Delta t$ [in reduced units]')
plt.ylabel('relative error at midpoint in %')
plt.title(r"$log$-$log$ plot of the relative error as a function of $\Delta t$")
plt.legend()

plt.savefig("error.pdf",dpi=400)
plt.show()
```

```python
""""
plotting the experimental data
"""

def plot_exp(file, text=''):
    
    # Open the text file
    with open(file, "r") as file:
        # Read each line of the file
        lines = file.readlines()

    # Initialize lists to store the time and temperature data
    times = []
    temperatures = [[] for _ in range(8)]

    # Loop over each line in the file
    for line in lines:
        # Split the line into values using the semicolon separator
        values = line.strip().split(";")
        
        # Check if the last value is empty
        if not values[-1]:
            values = values[:-1]
        
        # Extract the time and temperature data
        time = float(values[1])
        temp_values = [float(val) for val in values[2:]]
        
        # Append the data to the lists
        times.append(time)
        for i, temp in enumerate(temp_values):
            temperatures[i].append(temp)

    # Plot the temperature data
    plt.figure()
    for temp_data in temperatures:
        plt.plot(times, temp_data)
    plt.xlabel("Time [s]", fontsize=14)
    plt.ylabel("Temperature [°C]", fontsize=14)
    #plt.title("Coppper Bar: Temperature over Time ")
    # Add a text box to the top left of the plot
    plt.annotate("Copper Bar"+"\n"+text, (0.05, 0.95), xycoords="axes fraction", fontsize=12,
             ha="left", va="top", bbox=dict(boxstyle="round,pad=0.3", fc = 'yellow'
, alpha=0.2))

    #plt.savefig(text+".pdf", dpi=400)
    plt.show()

text1 = "(Without Fan)"
text2 = "(With Fan)"
file1 = "Cu_isol.txt"
file2 = "Cu_vent.txt"
plot_exp(file1,text1)
plot_exp(file2,text2)
```
