```python
import numpy as np
import matplotlib.pyplot as plt
# Specify the path to your ENCUT.dat file
file_path = 'ENCUT.dat'
file2 = 'ENCUT20Kpoints.dat'

E_20 = []
# Open the file and read each line
with open(file2, 'r') as file:
    for line in file:
        # Split the line into words
        words = line.split()

        #print(words)
        E_20.append(float(words[3]))
E_20 = np.array(E_20)

# Initialize empty lists to store the values
values_list = []
Kpoints = []
Encut = []
energy = []
# Open the file and read each line
with open(file_path, 'r') as file:
    for line in file:
        # Split the line into words
        words = line.split()
        #print(words)
        Kpoints.append(float(words[0]))
        Encut.append(float(words[1]))
        #energy.append(float(words[9][1:]))
        energy.append(float(words[4]))
# Extract variables
encut = np.array(Encut)
encut = encut[:7]
print(encut)
k_points = np.array(Kpoints)
energy = np.array(energy)
E_4 = energy[:7]
E_6 = energy[7:14]
E_8 = energy[14:21]
E_10 = energy[21:]



print("energy difference between k=8 and k=20:\n", np.abs(E_8-E_20)*1000/2)
print("energy difference between k=10 and k=20:\n", np.abs(E_10-E_20)*1000/2)
#print(E_4)
#print(E_6)
# Create a 3D plot
plt.plot(encut, E_4, 'o-',label='K=4')
plt.plot(encut, E_6, 'o-',label='K=6')
plt.plot(encut, E_8, 'o-',label='K=8') 
plt.plot(encut, E_10, 'o-',label='K=10')
plt.plot(encut, E_20, 'o-',label='K=20')


# Set labels
# Show the plot
plt.legend()
plt.ylim([-16.48, -16.46])
plt.show()


```
