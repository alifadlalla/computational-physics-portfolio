```python
import matplotlib.pyplot as plt
import numpy as np

# Specify the path to your data file
file_path = 'energy.dat'

# Read data from the file
with open(file_path, 'r') as file:
    data = file.read()

# Parse the data
lines = data.split('\n')
values = [line.split() for line in lines if line]
#print(values)
#print(values[0][0], values[0][8])

# Extracting values for x (distance) and y (energy)
x_values = [float(value[0]) for value in values]
y_values = [float(value[8][1:]) for value in values]

# Plotting the data
plt.plot(x_values, y_values, marker='o', linestyle='-')
plt.title('Energy vs Distance')
plt.xlabel(r'$d_{OH}$ $(A^o)$')
plt.ylabel('Energy (ev)')
plt.grid(True)
plt.savefig("energyH2O.pdf")
plt.show()

# Calculate binding energy (minimum energy)
min_energy = min(y_values)
print(f"Binding Energy: {min_energy} eV")
```

```python
def birch_murnaghan_energy(V, V0, B0, B0_prime, E0):
    """
    Calculate the energy using the Birch-Murnaghan equation of state.

    Parameters:
    - V: Volume
    - V0: Equilibrium volume
    - B0: Bulk modulus
    - B0_prime: Pressure derivative of bulk modulus
    - E0: Energy at equilibrium volume

    Returns:
    - Energy at the given volume
    """
    term1 = ((V0 / V) ** (2 / 3)) - 1
    term3 = (6 - (4 * (V0 / V) ** (2 / 3)))

    energy = E0 + (9 * V0 * B0 / 16) * (term1**3*B0_prime + term1**2*term3)

    return energy
```
