```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix

# Load Iris dataset
file_path = "iris.data"
data = pd.read_csv(file_path, header=None, delimiter=' ')  # Assuming space-separated data
input_data = data.iloc[:, :-1].values
output_data = data.iloc[:, -1].values

# Convert class labels to one-hot encoding
num_classes = len(np.unique(output_data))
output_data_one_hot = np.eye(num_classes)[output_data.astype(int)]

# Training parameters
learning_rate = 0.1
epochs = 50

# Initialize weights with small random values
weights = np.random.rand(input_data.shape[1] + 1, num_classes) * 0.2 - 0.1
print("Initial weights:", weights)

def perceptron_predict(inputs, weights):
    return np.argmax(np.dot(inputs, weights), axis=1)

for iter in range(epochs):
    for i in range(len(input_data)):
        target_output = output_data_one_hot[i]
        inputs = np.append(input_data[i], -1)
        g = np.dot(inputs, weights)
        y = np.argmax(g)
        error = target_output - np.eye(num_classes)[y]
        weights += learning_rate * np.outer(inputs, error)

    if (iter + 1) % 10 == 0:
        predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
        conf_matrix = confusion_matrix(output_data, predictions)
        accuracy = np.sum(np.diag(conf_matrix)) / np.sum(conf_matrix)
        print(f"Confusion Matrix at iteration {iter + 1}:\n{conf_matrix}")
        print(f"Accuracy at iteration {iter + 1}: {accuracy * 100:.2f}%")

# Final predictions and accuracy
final_predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
final_conf_matrix = confusion_matrix(output_data, final_predictions)
final_accuracy = np.sum(np.diag(final_conf_matrix)) / np.sum(final_conf_matrix)
print(f"Final Confusion Matrix:\n{final_conf_matrix}")
print(f"Final Accuracy: {final_accuracy * 100:.2f}%")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split

# Function to define the Perceptron for multi-class classification
def perceptron_predict(inputs, weights):
    return np.argmax(np.dot(inputs, weights))

# Function to one-hot encode class labels
def one_hot_encode(labels, num_classes):
    encoded_labels = np.zeros((len(labels), num_classes))
    for i, label in enumerate(labels):
        encoded_labels[i, label] = 1
    return encoded_labels

# Function to train the Perceptron for multi-class classification
def train_perceptron(input_data, output_data, learning_rate, epochs, num_classes):
    num_features = len(input_data[0])
    weights = np.random.rand(num_features + 1, num_classes)  # Initialize weights with small random values (including bias)

    for epoch in range(epochs):
        for i in range(len(input_data)):
            target_output = one_hot_encode([output_data[i]], num_classes)
            
            inputs = np.concatenate([input_data[i], [1]])  # Adding bias node with value 1
            predicted_output = perceptron_predict(inputs, weights)
            error = target_output - predicted_output
            
            weights += learning_rate * np.outer(inputs, error)

        # Print confusion matrix every few iterations
        if (epoch + 1) % 10 == 0:
            predictions = np.array([perceptron_predict(np.concatenate([input_data[i], [1]]), weights) for i in range(len(input_data))])
            cm = confusion_matrix(output_data, predictions)
            print(f"Iteration {epoch + 1} - Confusion Matrix:")
            print(cm)

    return weights

# Load Iris dataset
file_path = "iris.data"
data = pd.read_csv(file_path, header=None, delimiter=' ')  # Assuming space-separated data
input_data = data.iloc[:, :-1].values
output_data = data.iloc[:, -1].values

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(input_data, output_data, test_size=0.2, random_state=42)

# Training parameters
learning_rate = 0.1
epochs = 100
num_classes = 3

# Train the Perceptron for multi-class classification
weights = train_perceptron(X_train, y_train, learning_rate, epochs, num_classes)

# Measure final confusion matrix
final_predictions = np.array([perceptron_predict(np.concatenate([X_test[i], [1]]), weights) for i in range(len(X_test))])
final_cm = confusion_matrix(y_test, final_predictions)
print("\nFinal Confusion Matrix:")
print(final_cm)
```
