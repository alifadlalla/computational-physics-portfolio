```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix

# Load Iris dataset
file_path = "iris.data"
data = pd.read_csv(file_path, header=None, delimiter=' ')  # Assuming space-separated data
input_data = data.iloc[:, :-1].values
output_data = data.iloc[:, -1].values

# Convert class labels to one-hot encoding
num_classes = len(np.unique(output_data))
output_data_one_hot = np.eye(num_classes)[output_data.astype(int)]

# Training parameters
learning_rate = 0.1
epochs = 50

# Initialize weights with small random values
weights = np.random.rand(input_data.shape[1] + 1, num_classes) * 0.2 - 0.1
print("Initial weights:", weights)

def perceptron_predict(inputs, weights):
    return np.argmax(np.dot(inputs, weights), axis=1)

for iter in range(epochs):
    for i in range(len(input_data)):
        target_output = output_data_one_hot[i]
        inputs = np.append(input_data[i], -1)
        g = np.dot(inputs, weights)
        y = np.argmax(g)
        error = target_output - np.eye(num_classes)[y]
        weights += learning_rate * np.outer(inputs, error)

    if (iter + 1) % 10 == 0:
        predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
        conf_matrix = confusion_matrix(output_data, predictions)
        accuracy = np.sum(np.diag(conf_matrix)) / np.sum(conf_matrix)
        print(f"Confusion Matrix at iteration {iter + 1}:\n{conf_matrix}")
        print(f"Accuracy at iteration {iter + 1}: {accuracy * 100:.2f}%")

# Final predictions and accuracy
final_predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
final_conf_matrix = confusion_matrix(output_data, final_predictions)
final_accuracy = np.sum(np.diag(final_conf_matrix)) / np.sum(final_conf_matrix)
print(f"Final Confusion Matrix:\n{final_conf_matrix}")
print(f"Final Accuracy: {final_accuracy * 100:.2f}%")
```

```python
import numpy as np
import pandas as pd

# Load Iris dataset
file_path = "iris.data"
data = pd.read_csv(file_path, header=None, delimiter=' ')
input_data = data.iloc[:, :-1].values
output_data = data.iloc[:, -1].values

# Convert class labels to one-hot encoding
num_classes = len(np.unique(output_data))
output_data_one_hot = np.eye(num_classes)[output_data.astype(int)]

# Training parameters
learning_rate = 0.1
epochs = 50

# Initialize weights with small random values
weights = np.random.rand(input_data.shape[1] + 1, num_classes) * 0.2 - 0.1
print("Initial weights:", weights)

def perceptron_predict(inputs, weights):
    return np.argmax(np.dot(inputs, weights), axis=1)

# Confusion matrix function
def compute_confusion_matrix(actual, predicted, num_classes):
    conf_matrix = np.zeros((num_classes, num_classes), dtype=int)
    for a, p in zip(actual, predicted):
        conf_matrix[a, p] += 1
    return conf_matrix

# Accuracy calculation function
def compute_accuracy(conf_matrix):
    correct_predictions = np.sum(np.diag(conf_matrix))
    total_predictions = np.sum(conf_matrix)
    return correct_predictions / total_predictions

for iter in range(epochs):
    for i in range(len(input_data)):
        target_output = output_data_one_hot[i]
        inputs = np.append(input_data[i], -1)
        g = np.dot(inputs, weights)
        y = np.argmax(g)
        error = target_output - np.eye(num_classes)[y]
        weights += learning_rate * np.outer(inputs, error)

    if (iter + 1) % 10 == 0:
        predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
        conf_matrix = compute_confusion_matrix(output_data, predictions, num_classes)
        accuracy = compute_accuracy(conf_matrix)
        print(f"Confusion Matrix at iteration {iter + 1}:\n{conf_matrix}")
        print(f"Accuracy at iteration {iter + 1}: {accuracy * 100:.2f}%")

# Final predictions and accuracy
final_predictions = perceptron_predict(np.column_stack((input_data, np.ones(len(input_data)))), weights)
final_conf_matrix = compute_confusion_matrix(output_data, final_predictions, num_classes)
final_accuracy = compute_accuracy(final_conf_matrix)
print(f"Final Confusion Matrix:\n{final_conf_matrix}")
print(f"Final Accuracy: {final_accuracy * 100:.2f}%")
```
