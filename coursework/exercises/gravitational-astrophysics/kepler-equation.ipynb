```python
# ex 2
import numpy as np
M = 1  
eccentricities = [0.1, 0.7, 0.95]   

def scheme(M, e, var):
    E = M  #E0
    E_values = [E]
    for n in range(50):
        if var == 'fixed_point':
            E = M + e * np.sin(E)
        elif var=='Newton':
            E = E - (E - e * np.sin(E) - M) / (1 - e * np.cos(E))
        E_values.append(E)
    return E_values

print("Fixed-Points / Newton")
for e in eccentricities:
    E_fp = scheme(M, e, var= 'fixed_point' )[:10]
    E_N = scheme(M, e, var= 'Newton' )[:10]
    matrix = np.column_stack((E_fp,E_N))
    print(f"e={e}\n",matrix)
```

```python
import sympy as sp

# Define the variable e and n
e = sp.symbols('e')
n = sp.symbols('n', integer=True)

# Function to print Bessel function expansion for specific n
def Jn(n_value, order=5):
    return sp.besselj(n_value, n_value*e).series(e, 0, order).removeO()

# Print Bessel function expansions for n = 1 to n = 8
for n in range(1, 9):
    print(f"J_{n}(e) expansion:")
    Jn = sp.besselj(n, n*e).series(e, 0, 5).removeO()
    print(Jn)
```

```python
# Define the first few Bessel functions for small eccentricities (manual expansion)
J1 = e/2 - (e**3)/16  # Bessel function of first kind J1(e) to 3rd order
J2 = (e**2)/8 - (e**4)/192  # Bessel function of first kind J2(e) to 4th order
J3 = (e**3)/48  # Bessel function of first kind J3(e) to 3rd order
J4 = (e**4)/384  # Bessel function of first kind J4(e) to 4th order

# Now create the sum approximation with first 4 terms
E_approx = M + 2 * ((J1 * sp.sin(M)) + (J2 * sp.sin(2*M)) / 2 + (J3 * sp.sin(3*M)) / 3 + (J4 * sp.sin(4*M)) / 4)

# Perform Taylor expansion of the result to the 4th order in eccentricity
E_taylor_manual = E_approx.series(e, 0, 5).removeO()

E_taylor_manual
```
