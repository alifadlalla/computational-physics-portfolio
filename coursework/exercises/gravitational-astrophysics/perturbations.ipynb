```python
import sympy as sp

# Define the variable
e, M, theta, n, B, A, C, t = sp.symbols('e M theta n B A C t')
epsilon = (B - A) / C
order = 3 # if you want order 2 you have to use 2+1 for the code 

a_r = 1 + 2*sum([(sp.besselj(n, n*e)*sp.cos(n*M)).series(e, 0, order).removeO() for n in range(1, 9)])
# alternative way you can split into two steps
#a_r_series = 1 + 2 * sum([sp.besselj(k, k*e) * sp.cos(k*M) for k in range(1, 3)])
#a_r_expanded = sp.series(a_r_series, e, 0, order).removeO()


cos_f = (2*(1-e**2)/e)*sum([(sp.besselj(n, n*e)*sp.cos(n*M)).series(e, 0, order).removeO() for n in range(1, 9)])-e
sin_f = 2*(sp.sqrt(1-e**2))*sum([(sp.diff(sp.besselj(n, n*e),e)*sp.sin(n*M)/n).series(e, 0, order).removeO() for n in range(1, 9)])
```

```python
# Substitute the expansions into equation (5)
# Compute (a/r)^3
a_r_cubed = sp.series(a_r**3, e, 0, order).removeO()

# Substitute cos^2(f) and sin(f) * cos(f) for terms in the equation
cos_f_squared = sp.series(cos_f**2, e, 0, order).removeO()
sin_f_cos_f = sp.series(sin_f * cos_f, e, 0, order).removeO()

# Build equation (5) using expanded forms

equation_5_expanded = sp.simplify(
    sp.diff(theta, t, t) + (3 / 2) * n**2 * epsilon * a_r_cubed * (
        (2 * cos_f_squared - 1) * sp.sin(2 * theta) - 2 * sin_f_cos_f * sp.cos(2 * theta)
    )
) 

# Expand the equation to second order in eccentricity and remove higher-order terms
Theta_expression = sp.series(equation_5_expanded, e, 0, order).removeO()

# Display the final expanded equation
Theta_expression
```
