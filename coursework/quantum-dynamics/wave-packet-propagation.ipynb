## 2.1 System

In this practical work, we will study the time evolution of wave packets in different quantum systems using the split operator method, which is a numerical method used to solve the time-dependent Schrödinger equation.

Our system is described by a Hilbert space, $\mathcal{H}=L^2(\mathbb{R},dx)$, and a Hamiltonian, $\mathcal{H}=-\frac{\hbar^2}{2m}\frac{d^2}{dx^2} +V(x,t)$. We are interested in solving the Schrödinger equation, $i\hbar\dot{\psi}(t)= H(t)\psi(t)$, with an initial wave packet, $\psi(x,t=0)= \frac{1}{\sqrt2\pi}\int_{-\infty}^{+\infty} \psi_k(t=0)e^{ikx}dk$.

To solve the Schrödinger equation numerically, we divide the space and time into discrete partitions, {$x_0, ..., x_M$} with a step size of $\Delta x$, and {$t_0, ..., t_N$} with a step size of $\Delta t$. We will use the split operator method to compute the time evolution of the wave packet and study its behavior in different quantum systems.

## 2.2 The split operator method with DVR/FBR bases
The split operator method is a numerical method used to solve the time-dependent Schrödinger equation in quantum mechanics. This method is based on the fact that the kinetic energy operator, H_K, is diagonal in a Fourier basis representation (FBR), while the potential energy operator, V(x,t), is diagonal in a discrete variable representation (DVR).

To compute the time evolution of the wave function using this method, we first compute the matrix exponentials of H_K and V(x,t) in the two different bases. Then, we use the Baker-Campbell-Hausdorff formula to derive an integration scheme that consists of a series of steps involving matrix exponentials, Fourier transforms, and updates to the wave function at each time step.

This method can be used to study the time evolution of wave packets in various quantum systems, including the systems we will be studying in this practical work: a wave packet in a box and quantum carpet, a wave packet in front of a potential barrier, and a quantum Fermi accelerator. By solving the time-dependent Schrödinger equation using the split operator method, we can obtain the probability distribution of the wave packet at different times and understand how the wave packet evolves and disperses in space.

## 3.2 Wave packet in front of a potential barrier

In this section, we will consider a particle with mass $m=1$ a.u. in a region represented by $[0, x_{max}]$, where $x_{max} = 20$ a.u. In this region, a potential barrier has been erected from $x_0=10 \: a.u.$ to $x_1=11 \: a.u.$, with a height of $V_0=3 \: a.u.$.
$$
V(x) = \begin{cases} 
  0 & \text{if } x<x_0 \\ 
  V_0 & \text{if } x\in\,[x_0, x_1]\\ 
  0 & \text{if } x>x_1 \\ 
  \end{cases} 
$$





The initial state of the particle is a gaussian wave packet with standard deviation $\sigma_x=0.5$ a.u., centered at $x=5$ a.u., and with an initial momentum $k_0>0$, indicating that it is propagating to the right. The wave packet is given by:
<br>
<br>


$$
\psi(x,t=0)=\frac{1}{\sqrt{\sigma_{x}\sqrt{2\pi}}}e^{-\frac{(x-\frac{x_{0}}{2})^{2}}{4\sigma^{2}_{x}}}e^{ik_{0}x}
$$

<br>
    
We will study the dynamics of this wave packet as it propagates through the potential barrier.

#### Question 1: Compute $k_E=\frac{\sqrt{2mV_0}}{\hbar}$. What is the physical meaning of $\hbar K_E$ in classical dynamics?

```python
m = 1 #a.u.
V0 = 3 #a.u.
hbar = 1 #a.u.

KE = np.sqrt(2*m*V0)/hbar
print("K_E =", round(KE, 2), "a.u.")
```

$k_E \approx 2.45 \: a.u.$

The value of $k_E = \frac{\sqrt{2mV_0}}{\hbar}$ is known as the wave vector of the incident wave. In classical dynamics, $\hbar k_E$ is the minimum kinetic energy required for a particle to overcome the potential barrier with height $V_0$. In other words, it represents the minimum energy required for a particle to pass through the barrier, and is known as the "barrier height" or "barrier energy".

#### Question 2: After the construction of the system Hamiltonian, propagate the wave function during the time until tmax = 4 a.u. (we can choose ntime equal to 100), for different values of k0 below and above kE.

```python
# Potential barrier parameters
m = 1

x0 = 10
x1 = 11
xmax = 20

V0 = 3
sigmax = 0.5

tmin = 0
tmax = 4
Nstep = 100
ntime = 100
DeltatX = xmax/Nstep
Deltat = tmax/ntime
Xd = np.linspace(0, xmax, Nstep+1)

# Potential
def V_barrier(x):
    if x >= x0 and x <= x1:
        return V0
    else:
        return 0


# Hamiltonian 
HK_barrier = HK(m, xmax, Nstep)
H_barrier = H_t_ind(V_barrier, HK_barrier)

# psi
def psi(K0=KE):
    # Computing psi

    expH_barrier = LA2.expm(-1j*H_barrier*Deltat) # time_independant H

    psi = np.zeros((Nstep+1, ntime+1), dtype=complex)

    # psi at t=0
    for i in range(Nstep+1):
        psi[i, 0] = (1/np.sqrt(sigmax * np.sqrt(2*pi))) * np.exp(-pow(Xd[i] - x0/2, 2) / (4*sigmax**2)) * np.exp(1j*K0*Xd[i])

    # psi at t
    for i in range(ntime):
        psi[:, i+1] = expH_barrier@psi[:, i]
        
    return psi
    
```

## 3.3 Quantum Fermi accelerator
<br>
We consider a particle with mass $m=1 \: a.u.$ in a box $[0,x_{max}]$ with $x_{max}=20 \: a.u.$ A potential barrier with time-dependent width is erected on the right side of the box:

$$
V(x, t) = \begin{cases} 
  0 & \text{if } x<x_{max}-\frac{(x_{max}-x_{min)}}{2}(1-cos(\omega t))\\ 
  V_0 & \text{if } x\geq x_{max}-\frac{(x_{max}-x_{min)}}{2}(1-cos(\omega t))\\ 
  \end{cases}
$$

where $V_0=10^{15}$ a.u., $\omega=\frac{2\pi}{\tau}$, and $x_{\text{min}}=\frac{3}{4}x_{\text{max}}$. 

At the initial time, the particle is in an eigenvector of $H(t=0)$:

$$\psi(x,t=0)=\sqrt{\frac{2}{x_{\text{max}}}}\sin\left(\frac{p\pi x}{x_{\text{max}}}\right)$$

where $p\in\mathbb{N}$.

#### Question 1: After the construction of the system Hamiltonian, propagate the wave packet during the time until tmax = 30 a.u. (we can choose ntime equal to 400), for different values of $\tau$ included between 0.5 and 30 a.u., for p = 1, 2 or 3.

```python
# Quantum Fermi accelerator parameters
m = 1

xmax = 20
xmin = (3/4)*xmax

V0 = 10**15

tmin = 0
tmax = 30
Nstep = 100
ntime = 400
DeltatX = xmax/Nstep
Deltat = tmax/ntime
Xd = np.linspace(0, xmax, Nstep+1)


def psi_Fermi(tau=0.5, p=1):
    
    omega = 2*pi/tau
    
    def V_Fermi(x, t):
        if x < xmax - 0.5*(xmax - xmin)*(1 - cos(omega*t)):
            return 0
        else:
            return V0


    psi = np.zeros((Nstep+1, ntime+1), dtype=complex)

    # psi at t=0
    for i in range(Nstep+1):
        psi[i, 0] = np.sqrt(2/xmax) * sin(p*pi*Xd[i]/xmax)


    # Hamiltonian DVR
    HK_Fermi = HK(m, xmax, Nstep)

    # time_dependant H
    expHK = LA2.expm(-1j*HK_Fermi*Deltat) # time_independant H

    for i in range(ntime):
        ExpV = np.diag([np.exp(-1j*V_Fermi(Xd[j], i*Deltat)*Deltat/2) for j in range(Nstep+1)])
        ExpH = ExpV@expHK@ExpV
        psi[:, i+1] = ExpH@psi[:, i]
    return psi


psi = psi_Fermi(tau=0.5, p=1)
```

For the three simulations shown above, we can see that as the value of the $\tau$ increases, the intensity of the reflected wave packet from the potential barrier decreases and the direction of propagation becomes more continuous. This behavior can be understood by considering the energy of the wave packet and the shape of the potential barrier.

As the value of tau increases, the width of the potential barrier decreases, effectively lowering the barrier height.
This can be explained by the fact that the width of the potential barrier in the quantum Fermi accelerator model is given by the expression $x_{max}-\frac{(x_{max}-x_{min)}}{2}(1-cos(\omega t))$, where $x_{max}$ and $x_{min}$ are the maximum and minimum width of the barrier, respectively, and $\omega=\frac{2\pi}{\tau}$.
From this expression, we can see that as the value of $\tau$ increases, the value of $\omega$ decreases, and therefore the cosine term in the expression oscillates more slowly, resulting in a decrease in the width of the potential barrier. This is why we observe that as the value of tau increases, the width of the potential barrier decreases.

For a given initial momentum of the wave packet, a lower barrier height means that it is more likely for the wave packet to pass through the barrier rather than being reflected. This is similar to the phenomenon of tunneling, where particles can exhibit wave-like behavior and pass through barriers that would be impenetrable for classical particles.

Additionally, we can see that the number of nodes in the initial wave packet changes as we vary the value of the parameter p. This reflects the fact that the initial wave packet is an eigenvector of the Hamiltonian at t=0, and the number of nodes in the wave function corresponds to the energy of the particle. As the value of p increases, the energy of the particle also increases, which can affect the probability of transmission through the potential barrier.
