```python
import numpy as np
import matplotlib.pyplot as plt

# Chirikov Standard Map function
def chirikov_standard_map(I, theta, K):
    I_next = (I + K * np.sin(theta)) % (2 * np.pi)
    theta_next = (theta + I_next) % (2 * np.pi)
    return I_next, theta_next

# Function to perform the numerical simulation
def simulate_chirikov_standard_map(K):
    # Initial conditions
    I_0 = 0.5
    theta_0 = 0.0

    # Number of iterations
    num_iterations = 1000

    # Arrays to store results
    I_values = np.zeros(num_iterations)
    theta_values = np.zeros(num_iterations)

    # Initial conditions
    I_values[0] = I_0
    theta_values[0] = theta_0

    # Perform simulation
    for i in range(1, num_iterations):
        I_values[i], theta_values[i] = chirikov_standard_map(I_values[i - 1], theta_values[i - 1], K)

    return I_values, theta_values

# Function to find K_c
def find_critical_K():
    K_values = np.linspace(0, 1, 1000)  # Adjust the range based on your needs
    domain_coverage = []

    for K in K_values:
        I_values, _ = simulate_chirikov_standard_map(K)
        domain_coverage.append(np.ptp(I_values))  # Peak-to-peak distance

    # Find the critical K_c where the domain coverage is close to 2*pi
    K_c = K_values[np.argmax(np.array(domain_coverage) >= 2 * np.pi)]

    return K_c

# Find and print the critical K_c
critical_K = find_critical_K()
print("Critical K_c:", critical_K)
```

```python
import numpy as np

# Function implementing the Chirikov standard map
def chirikov_map(I, theta, K):
    """
    Implements the Chirikov standard map:
    I_{n+1} = I_n + K * sin(theta_n) mod(2*pi)
    theta_{n+1} = theta_n + I_{n+1} mod(2*pi)

    Args:
        I: Current value of the action variable.
        theta: Current value of the angle variable.
        K: Parameter of the map.

    Returns:
        A tuple containing the new values of I and theta.
    """
    I_new = (I + K * np.sin(theta)) % (2 * np.pi)
    theta_new = (theta + I_new) % (2 * np.pi)
    return I_new, theta_new

# Function to check if a trajectory is chaotic
def is_chaotic(I, theta, K, max_iter=1000):
    """
    Checks if a trajectory is chaotic by iterating the map and
    checking if I explores the full domain.

    Args:
        I: Initial value of the action variable.
        theta: Initial value of the angle variable.
        K: Parameter of the map.
        max_iter: Maximum number of iterations.

    Returns:
        True if the trajectory is chaotic, False otherwise.
    """
    for _ in range(max_iter):
        I, theta = chirikov_map(I, theta, K)
        if I < 0.01 or I > 2 * np.pi - 0.01:  # Check if I spans the full domain
            return True
    return False

# Initial conditions
I0 = 0.5
theta0 = 0.0

# Search for the critical value of K using a bisection method
Kc_min = 0.0
Kc_max = 10.0
Kc_tol = 1e-6
while Kc_max - Kc_min > Kc_tol:
    Kc_mid = (Kc_min + Kc_max) / 2
    if is_chaotic(I0, theta0, Kc_mid):
        Kc_min = Kc_mid
    else:
        Kc_max = Kc_mid

print(f"Estimated critical value of K: {Kc_min:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def chirikov_map(I, theta, K):
    I_next = (I + K * np.sin(theta)) % (2 * np.pi)
    theta_next = (theta + I_next) % (2 * np.pi)
    return I_next, theta_next

def lyapunov_exponent(I_init, theta_init, K, num_iterations):
    delta_I = 1e-10  # small perturbation
    I, theta = I_init, theta_init
    I_perturbed = I_init + delta_I
    sum_log_delta_I = 0
    for _ in range(num_iterations):
        I_next, theta_next = chirikov_map(I, theta, K)
        I_perturbed_next, _ = chirikov_map(I_perturbed, theta, K)
        delta_I = np.abs(I_perturbed_next - I_next)
        sum_log_delta_I += np.log(delta_I)
        I, theta = I_next, theta_next
        I_perturbed = I_perturbed_next
    return sum_log_delta_I / num_iterations

M = 100  # number of initial conditions
I_initial = np.linspace(0, 2*np.pi, M)
theta_initial = np.linspace(0, 2*np.pi, M)

K_values = np.linspace(0,4, 100)  # range of K values to test
num_iterations = 1000  # number of iterations for each K value

lyapunov_exponents = np.zeros((M, M, len(K_values)))

for i, I_init in enumerate(I_initial):
    for j, theta_init in enumerate(theta_initial):
        lyapunov_exponents[i, j, :] = [lyapunov_exponent(I_init, theta_init, K, num_iterations) for K in K_values]

plt.imshow(lyapunov_exponents.mean(axis=(0, 1)), extent=[K_values.min(), K_values.max(), 0, 1], aspect='auto')
plt.colorbar(label='Lyapunov exponent')
plt.xlabel("$K$")
plt.ylabel("Mean Lyapunov exponent")
plt.show()
```
