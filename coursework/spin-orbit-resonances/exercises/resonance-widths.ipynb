```python
import numpy as np

# Constants
GMs = 37931206.2  # Saturn's gravitational parameter (km^3/s^2)
J2 = 1.46965e-2  # Saturn's second zonal harmonic
R = 60330  # Saturn's mean equatorial radius (km)
GM1 = 2.50349  # Gravitational parameter for Mimas (km^3/s^2)
GM3 = 41.21353  # Gravitational parameter for Tethys (km^3/s^2)
a1 = 1.86e5  # Semi-major axis of Mimas (km)
a3 = 2.95e5  # Semi-major axis of Tethys (km)
e1 = 2.0e-2  # Eccentricity of Mimas
e3 = 2.0e-4  # Eccentricity of Tethys
I1 = np.radians(1.6)  # Inclination of Mimas (rad)
I3 = np.radians(1.1)  # Inclination of Tethys (rad)

# Mean motions (approximation using Kepler's third law)
def mean_motion(GM, a):
    return np.sqrt(GM / a**3)

n1 = mean_motion(GMs + GM1, a1)
n3 = mean_motion(GMs + GM3, a3)

# Calculate precession rates
precession_factor = (3/2) * J2 * (R / a1)**2

omega_dot_1 = precession_factor * n1  # Mimas
omega_dot_3 = precession_factor * n3  # Tethys

Omega_dot_1 = -precession_factor * n1
Omega_dot_3 = -precession_factor * n3

print("Mimas")
print("r$\omega$",omega_dot_1)
# Resonance arguments
p_q_values = [
    (1, 1, 1, 0, 0, 0),
    (2, 2, 1, 1, 0, 0),
    (1, 1, 0, 1, 0, 0),
    (2, 2, 0, 0, 0, 2),
    (2, 2, 0, 0, 1, 1),
    (2, 2, 0, 0, 2, 0),
]

def resonance_location(p, q, q1, q2, q3, q4):
    phi_dot = p * n1 - (p + q) * n3 + q1 * omega_dot_1 + q2 * omega_dot_3 + q3 * Omega_dot_1 + q4 * Omega_dot_3
    return phi_dot

locations = [
    resonance_location(p, q, q1, q2, q3, q4)
    for p, q, q1, q2, q3, q4 in p_q_values
]

# Compute resonance widths
# Laplace coefficients
b_values = {
    "b(1,1/2)": 0.75691,
    "b(2,1/2)": 0.365372,
    "b(3,1/2)": 0.193966,
    "b(3,3/2)": 2.60382
}

A_coeffs = []
for i, (p, q, q1, q2, q3, q4) in enumerate(p_q_values):
    if i == 0:  # Example for phi1 calculation
        A = (
            (6 * n1**2 * (a1 / a3) * GM3 / GMs) + (24 * n3**2 * GM1 / GMs)
        ) * e1 * (4 + 0.63 * 1.75599) * b_values["b(2,1/2)"]
        A_coeffs.append(A)

widths = [np.sqrt(A) for A in A_coeffs]

# Output results
print("Resonance Locations (phi_dot = 0):")
for i, location in enumerate(locations):
    print(f"Resonance {i + 1}: phi_dot = {location:.6e}")

print("\nResonance Widths:")
for i, width in enumerate(widths):
    print(f"Resonance {i + 1}: Width = {width:.6e}")
```
